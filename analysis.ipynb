```python
autos.columns = [re.sub(r"([a-z])([A-Z])", r"\1_\2", col).lower() for col in autos.columns]
```

```python
autos.info()
```

```python
autos = autos.rename(columns={
    "year_of_registration": "registration_year",
    "kilometer": "odometer_km",
})
```

```python

autos = autos[autos['offer_type'] == 'Angebot']
```

```python
autos.columns
```

```python
autos = autos.drop(["nr_of_pictures", "seller", "offer_type", "abtest"], axis=1)
```

```python
autos.shape
```

```python
print(autos['vehicle_type'].value_counts())
print(autos['gearbox'].value_counts())
print(autos['not_repaired_damage'].value_counts())
print(autos['fuel_type'].value_counts())
```

```python
autos["gearbox"] = autos["gearbox"].replace({"manuell": "manual", "automatik": "automatic"})
```

```python
autos["vehicle_type"] = autos["vehicle_type"].replace({"limousine": "sedan", "kleinwagen": "small car", "kombi": "station wagon", "cabrio": "convertible", "andere": "other"})
```

```python
print(autos['vehicle_type'].value_counts())
print(autos['gearbox'].value_counts())
```

```python
autos = autos.rename(columns={"not_repaired_damage": "has_unrepaired_damage"})
```

```python
autos["has_unrepaired_damage"] = autos["has_unrepaired_damage"].replace({"nein": "no", "ja": "yes"})
```

```python
autos["fuel_type"] = autos["fuel_type"].replace({"benzin": "petrol", "elektro": "electric", "andere": "other"}) 
```

```python
print(autos['vehicle_type'].value_counts())
print(autos['gearbox'].value_counts())
print(autos['has_unrepaired_damage'].value_counts())
print(autos['fuel_type'].value_counts())
```
